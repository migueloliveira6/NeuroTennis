```python
import pandas as pd
import re

def normalize_name(name):
    """
    Normaliza nomes removendo informações extras como país, rankings, qualifiers, etc.
    
    Args:
        name (str): Nome original com informações extras
        
    Returns:
        str: Nome normalizado
    """
    if pd.isna(name) or name == "":
        return name
    
    # Remove qualifiers e wildcards no início: (Q), (WC), (LL), (2/WC), etc.
    name_clean = re.sub(r'^\([^)]*\)', '', name)
    
    # Remove tudo que está entre colchetes [...]
    name_clean = re.sub(r'\[.*?\]', '', name_clean)
    
    # Remove números entre parênteses (rankings)
    name_clean = re.sub(r'\(\d+\)', '', name_clean)
    
    # Remove caracteres especiais e símbolos extras
    name_clean = re.sub(r'[|,].*$', '', name_clean)
    
    # Remove espaços extras e limpa
    name_clean = re.sub(r'\s+', ' ', name_clean).strip()
    
    return name_clean

def normalize_dataset_names(df):
    """
    Normaliza os nomes nas colunas winner_name e loser_name
    
    Args:
        df (pandas.DataFrame): DataFrame com os dados
        
    Returns:
        pandas.DataFrame: DataFrame com nomes normalizados
    """
    df_normalized = df.copy()
    
    # Normaliza as colunas de nomes
    if 'winner_name' in df_normalized.columns:
        df_normalized['winner_name'] = df_normalized['winner_name'].apply(normalize_name)
    
    if 'loser_name' in df_normalized.columns:
        df_normalized['loser_name'] = df_normalized['loser_name'].apply(normalize_name)
    
    return df_normalized
```

```python
# Carregue seu dataset
df = pd.read_csv('D:/projetos/Tenis ML-AI/tennis_2025_data_v3.csv')

# Normalize os nomes
df_limpo = normalize_dataset_names(df)

# Salve o resultado
df_limpo.to_csv('dataset_2025_normalizado_v3.csv', index=False)
```
